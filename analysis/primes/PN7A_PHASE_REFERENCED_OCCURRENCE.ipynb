## Context & Methods

The direct survivor share is mapped to the ARA diameter by `x = 2(1-S)` and to the fixed upper circle branch by `theta = acos(2S-1)`. The independent occurrence reading compares exposure-corrected removal hazards in the right and left halves of the raw ordered interval: `a_O = (h_R-h_L)/(h_R+h_L)`. R9-R11 occurrence is interpolated onto one common adult-theta grid without a fitted shift, sign, rotation or smoother.


## Data

The aggregate packet contains exact R7-R11 p29-conditioned candidate and adjacent-edge exposures crossed with 24 normalized later-prime gate cells and 64 equal ordered position bins. These are already-open development windows; p31/R12 remains protected.


```python
import json
from pathlib import Path
import numpy as np
HERE = Path.cwd()
z = np.load(HERE / 'PN7A_PHASE_REFERENCED_OCCURRENCE_AGGREGATES.npz')
for rung in range(7, 12):
    candidate = int(z[f'r{rung}__candidate_exposure'].sum())
    terminal = int(z[f'r{rung}__candidate_stage_position'][24].sum())
    closes = int(z[f'r{rung}__candidate_stage_position'].sum()) == candidate
    print(f'R{rung}: candidates={candidate:,}; terminal={terminal:,}; partition_closes={closes}')
```

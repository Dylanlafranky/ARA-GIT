## Adult periods and the 1.0 ridge

The median exact joint periods are compared across decimal rungs. Consecutive growth multipliers are then normalised together to TE-ARA = 2.

```python
J = [row['geometry']['median_joint_period'] for row in dev['scales']]
J.append(target['target']['geometry']['median_joint_period'])
G = [J[i+1] / J[i] for i in range(3)]
ridge_open = (2*G[0]/(G[0]+G[1]), 2*G[1]/(G[0]+G[1]))
ridge_fresh = (2*G[1]/(G[1]+G[2]), 2*G[2]/(G[1]+G[2]))
{'J': J, 'G': G, 'open_ridge': ridge_open, 'fresh_ridge': ridge_fresh}
```

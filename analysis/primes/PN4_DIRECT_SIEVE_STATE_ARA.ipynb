# PN4 direct sieve-state ARA

## TL;DR

On already opened R6-R9 sieve data, the direct ARA survivor/release path and the explicit candidate/pair relation transfer most of the large deviation from independent sieving. They beat independence consistently, but the frozen raw multiplicative-ratio control is slightly better. The local three-point ARA stencil loses. This is useful cross-scale structure, not an ARA-only prime prediction result.

## Context & Methods

The source is the exact PN3A smallest-factor/death-gate archive. Each rung is divided into 24 equal normalized log-gate cells. The direct coordinate is $x=2(1-S)$, and the pair relation is $J=\log(S_e/S_c^2)$. R6-R8 supply prior-rung paths; R9 is held out by this code path but was already opened historically. Full definitions and decision rules are in `PN4_DIRECT_SIEVE_STATE_ARA_PROTOCOL.md`.

```python
import csv, json
from pathlib import Path

HERE = Path.cwd()
results = json.loads((HERE / 'PN4_DIRECT_SIEVE_STATE_RESULTS.json').read_text(encoding='utf-8'))
validation = json.loads((HERE / 'PN4_DIRECT_SIEVE_STATE_VALIDATION.json').read_text(encoding='utf-8'))
rows = list(csv.DictReader((HERE / 'PN4_DIRECT_SIEVE_STATE_PATHS.csv').open(encoding='utf-8')))
print(results['test_id'])
print('validation:', validation['checks_passed'], '/', validation['checks_total'], 'passed')
print('path rows:', len(rows))
```

## Data

The four paths contain candidate and adjacent-pair survivor counts at every fixed cell endpoint, their 0-2 ARA coordinates, the independent sieve products and the explicit coupling relation. No Fourier, SVD, NMF or future death-gate label is used as a predictor.

```python
for rung, summary in results['rung_summaries'].items():
    print(rung.upper(), 'candidate S=', f"{summary['candidate_terminal_survival']:.9f}",
          'edge S=', f"{summary['edge_terminal_survival']:.9f}",
          'J=', f"{summary['terminal_coupling_j']:.6f}")
```

## Results

```python
scores = results['transfers']['r8_to_r9']['scores']
for entity in ('candidate', 'edge'):
    print('\n' + entity.upper())
    for model, metric in scores[entity].items():
        print(f"{model:38s} loss={metric['log_loss_bits_per_at_risk_event']:.10f}  RMSE={metric['survival_rmse']:.7f}  terminal error={100*metric['terminal_absolute_relative_error']:.4f}%")
print('\nCRITERIA')
for key, value in results['criteria'].items():
    print(key, '=', value)
```

```python
terminal = results['terminal_established_comparators']
print('Mertens/PNT candidate terminal error:', f"{100*terminal['candidate']['absolute_relative_errors']['mertens_pnt_factor_prediction']:.4f}%")
print('Mertens/PNT-squared edge terminal error:', f"{100*terminal['edge']['absolute_relative_errors']['mertens_pnt_factor_squared_prediction']:.4f}%")
```

## Takeaways

1. The adult survivor/release deformation is strongly scale-transferable from R8 to R9 and repeats in direction from R7 to R8.
2. The predeclared ARA additive transfer is not the best model: multiplicative raw-ratio transfer has slightly lower path log loss and much lower terminal error.
3. The pair relation $J$ is informative enough to greatly improve on pair independence, but again the raw edge ratio is slightly better.
4. The local three-point ARA stencil does not beat both Home and independence.
5. The winning raw ratio is a promising candidate for a future, separately frozen multiplicative rung-coupling law. It cannot be relabelled as an ARA win after seeing this result.
6. Established Mertens/PNT terminal scaling remains exceptionally strong, so the present result is compatible with known finite-sieve convergence.

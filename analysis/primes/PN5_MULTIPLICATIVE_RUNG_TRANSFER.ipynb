# PN5 fresh R10 multiplicative rung transfer

## TL;DR

Predictions were frozen and hashed before constructing the 100-million-integer R10 target. The multiplicative ARA/raw-equivalent rung rule predicts candidate terminal survival to 0.0192% and adjacent-pair survival to 0.1310%. It decisively beats independence and the previous additive rule. The declared Buchstab asymptotic controls retain slightly better full-path log loss, and the pair relation J fails the stricter nearest-rung recurrence criterion.

## Context & Methods

The target is `[10,000,000,000,10,100,000,000)`. The frozen candidate rule preserves `k=log(S/S_independent)` from R9; the pair rule additionally preserves `J=log(S_edge/S_candidate^2)`. Twenty-four fixed normalized log-gate cells were used. The target was built exactly with a segmented Eratosthenes sieve only after the prediction packet hash was verified.

### Key Assumptions

The 24-cell coarse-graining is fixed from PN4. The Buchstab path is an asymptotic midpoint control for this short interval, not an exact finite-window theorem. The primary ARA model is algebraically identical to raw multiplicative-ratio transfer.

```python
import csv, json
from pathlib import Path
HERE = Path.cwd()
results = json.loads((HERE / 'PN5_MULTIPLICATIVE_RUNG_RESULTS.json').read_text(encoding='utf-8'))
validation = json.loads((HERE / 'PN5_FRESH_R10_VALIDATION.json').read_text(encoding='utf-8'))
rows = list(csv.DictReader((HERE / 'PN5_MULTIPLICATIVE_RUNG_PATHS.csv').open(encoding='utf-8')))
print(results['test_id'])
print('validation:', validation['checks_passed'], '/', validation['checks_total'])
print('path rows:', len(rows))
```

## Takeaways

1. The multiplicative/log-ratio rung law prospectively transfers with very high terminal accuracy.
2. It clearly supersedes the previously frozen additive ARA residual rule for this sieve path.
3. Buchstab remains slightly better on full-path log loss, so PN5 does not establish a new prime law.
4. Candidate k shows nearest-rung recurrence, but pair J does not; the pair correction appears to oscillate or converge non-monotonically.
5. The observed curve remains visually arc-like, but Buchstab shares the same bowed envelope. Circular appearance alone therefore does not distinguish the explanations.
6. Because the ARA primary is algebraically the same as a raw ratio, PN5 confirms recurrence and an interpretable decomposition, not ARA-unique information.

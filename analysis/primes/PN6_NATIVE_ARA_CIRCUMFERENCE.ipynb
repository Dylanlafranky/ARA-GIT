## Context & Methods

The direct survivor share is mapped to the ARA diameter by `x=2(1-S)`, then to the fixed upper circle branch by `theta=acos(2S-1)`. One shared phase-increment withdrawal factor is learned from R8-R10 and continued once to R11. No established prime-density law or spectral decomposition enters the native prediction.

```python
import json
from pathlib import Path

HERE = Path.cwd()
results = json.loads((HERE / 'PN6_NATIVE_ARA_RESULTS.json').read_text(encoding='utf-8'))
frozen = json.loads((HERE / 'PN6_NATIVE_ARA_FROZEN_PREDICTIONS.json').read_text(encoding='utf-8'))
target = json.loads((HERE / 'PN6_R11_TARGET_AGGREGATES.json').read_text(encoding='utf-8'))
validation = json.loads((HERE / 'PN6_NATIVE_ARA_VALIDATION.json').read_text(encoding='utf-8'))
print('Test:', results['test_id'])
print('Frozen rho:', results['frozen_rho_shared'])
print('Prediction hash preserved:', results['freeze_integrity']['matched'])
print('Independent validation:', validation['summary'])
```

## Data

The target is the complete interval `[100,000,000,000, 101,000,000,000)`. Exact segmented-sieve construction records the first later prime gate removing every p29-conditioned candidate and adjacent edge.

```python
for identity in ('candidate', 'edge'):
    counts = results['counts'][identity]
    share = counts['terminal_survivors'] / counts['initial']
    print(f"{identity:9s} initial={counts['initial']:,} terminal={counts['terminal_survivors']:,} share={share:.9f}")
print('Accounting:', results['target_accounting_checks'])
```

## Results

The primary circle path is compared only with native controls in this notebook. Lower log loss and RMSE are better.

```python
models = ('home_r10', 'direct_log_rung', 'circle_shared_rho_primary')
for identity in ('candidate', 'edge'):
    print('\n' + identity.upper())
    print('model                          logloss      survival_RMSE phase_RMSE terminal_error')
    for model in models:
        s = results['scores'][identity][model]
        print(f"{model:30s} {s['log_loss_bits_per_at_risk_event']:.9f} {s['survival_rmse']:.7f}       {s['phase_rmse_radians']:.7f}   {100*s['terminal_absolute_relative_error']:.4f}%")
print('\nObserved withdrawal:', results['observed_rho'])
```
